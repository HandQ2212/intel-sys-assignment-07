# Chương 3: Mạng Tích chập (CNN) - Framework: Keras - Dataset: CIFAR-10


```
import numpy as np
from tensorflow.keras.datasets import cifar10
(X_train, y_train), (X_test, y_test) = cifar10.load_data()
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
print(f"Loaded CIFAR-10 Keras: X_train {X_train.shape}")
```

```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Flatten, Dense
model = Sequential([Conv2D(16, (3,3), input_shape=(32, 32, 3)), Flatten(), Dense(10)])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
print('Keras CNN Model built')
```

```
import os
os.makedirs('../../models', exist_ok=True)
model_path = '../../models/CNN_Keras_CIFAR-10'
model.save(model_path + '.h5')
print('Saved Keras model to', model_path + '.h5')
```
