# Chương 3: Mạng Tích chập (CNN) - Framework: Keras - Dataset: MNIST


```
import numpy as np
from tensorflow.keras.datasets import mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
X_test = X_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0
print(f"Loaded MNIST Keras: X_train {X_train.shape}")
```

```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Flatten, Dense
model = Sequential([Conv2D(16, (3,3), input_shape=(28, 28, 1)), Flatten(), Dense(10)])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
print('Keras CNN Model built')
```

```
import os
os.makedirs('../../models', exist_ok=True)
model_path = '../../models/CNN_Keras_MNIST'
model.save(model_path + '.h5')
print('Saved Keras model to', model_path + '.h5')
```
