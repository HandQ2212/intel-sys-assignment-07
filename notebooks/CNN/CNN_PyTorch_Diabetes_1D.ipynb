# Chương 3: Mạng Tích chập (CNN) - Framework: PyTorch - Dataset: Diabetes_1D


```
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
df = pd.read_csv('../../data/diabetes.csv')
X = df.drop('Outcome', axis=1).values
y = df['Outcome'].values
X = StandardScaler().fit_transform(X)
X_train = X.reshape(X.shape[0], X.shape[1], 1)
print(f"Loaded Diabetes 1D dataset: X_train {X_train.shape}")
```

```
import torch.nn as nn
class CNN_Torch(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Conv2d(3 if 'Diabetes_1D'=='CIFAR-10' else 1, 16, 3)
model = CNN_Torch()
print('PyTorch CNN Model initialized')
```

```
import os
os.makedirs('../../models', exist_ok=True)
model_path = '../../models/CNN_PyTorch_Diabetes_1D'
import torch
torch.save(model.state_dict(), model_path + '.pt')
print('Saved PyTorch model to', model_path + '.pt')
```
