# Chương 3: Mạng Tích chập (CNN) - Framework: Keras - Dataset: Diabetes_1D


```
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
df = pd.read_csv('../../data/diabetes.csv')
X = df.drop('Outcome', axis=1).values
y = df['Outcome'].values
X = StandardScaler().fit_transform(X)
X_train = X.reshape(X.shape[0], X.shape[1], 1)
print(f"Loaded Diabetes 1D dataset: X_train {X_train.shape}")
```

```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, Flatten, Dense
model = Sequential([Conv1D(16, 3, activation='relu', input_shape=(21,1)), Flatten(), Dense(1)])
model.compile(optimizer='adam', loss='binary_crossentropy')
print('Keras CNN 1D Model built')
```

```
import os
os.makedirs('../../models', exist_ok=True)
model_path = '../../models/CNN_Keras_Diabetes_1D'
model.save(model_path + '.h5')
print('Saved Keras model to', model_path + '.h5')
```
