# Chương 4: Mạng Hồi quy (RNN) - Framework: Keras - Dataset: Gold_Price


```
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('../../data/gold_price.csv')
# Drop NA just in case
df = df.dropna()
data = df['Price' if 'Price' in df.columns else df.columns[1]].values.reshape(-1, 1)

scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)

def create_sequences(data, seq_length):
    xs, ys = [], []
    for i in range(len(data)-seq_length):
        xs.append(data[i:i+seq_length])
        ys.append(data[i+seq_length])
    return np.array(xs), np.array(ys)

X_train, y_train = create_sequences(data_scaled[:int(len(data)*0.8)], 30)
X_test, y_test = create_sequences(data_scaled[int(len(data)*0.8):], 30)
print(f"Loaded Gold_Price dataset: X_train {X_train.shape}")
```

```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
model = Sequential([SimpleRNN(16, input_shape=(30, 1)), Dense(1)])
model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=2, batch_size=32)
print('Keras RNN Model trained')
```

```
import os
os.makedirs('../../models', exist_ok=True)
model_path = '../../models/RNN_Keras_Gold_Price'
model.save(model_path + '.h5')
print('Saved Keras model to', model_path + '.h5')
```
