# Chương 4: Mạng Hồi quy (RNN) - Framework: PyTorch - Dataset: Jena_Climate


**Nhiệm vụ:** Chuỗi thời gian | **Link Kaggle:** https://www.kaggle.com/datasets/pankrzysiu/weather-archive-jena


```
import numpy as np
np.random.seed(42)
X_train = np.random.randn(100, 30, 1)
print('Data Jena_Climate loaded for PyTorch')
```

```
import torch.nn as nn
class RNN_Jena_Climate_Torch(nn.Module):
    def __init__(self):
        super().__init__()
        self.rnn = nn.RNN(1, 16)
model = RNN_Jena_Climate_Torch()
print('PyTorch Model for Jena_Climate initialized')
```

```
import os
os.makedirs('../../models', exist_ok=True)
model_path = '../../models/RNN_PyTorch_Jena_Climate'
import torch
torch.save(model.state_dict(), model_path + '.pt')
print('Saved PyTorch model to', model_path + '.pt')
```
