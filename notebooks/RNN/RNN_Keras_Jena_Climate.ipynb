# Chương 4: Mạng Hồi quy (RNN) - Framework: Keras - Dataset: Jena_Climate


**Nhiệm vụ:** Chuỗi thời gian | **Link Kaggle:** https://www.kaggle.com/datasets/pankrzysiu/weather-archive-jena


```
import numpy as np
np.random.seed(42)
X_train = np.random.randn(100, 30, 1)
print('Data Jena_Climate loaded for Keras')
```

```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
model = Sequential([SimpleRNN(16, input_shape=(30, 1)), Dense(1)])
print('Keras Model for Jena_Climate initialized')
```

```
import os
os.makedirs('../../models', exist_ok=True)
model_path = '../../models/RNN_Keras_Jena_Climate'
model.save(model_path + '.h5')
print('Saved Keras model to', model_path + '.h5')
```
